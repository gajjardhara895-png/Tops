```python
# 1. Create a Python class called Product with a private attribute _price. Initialize _price in the constructor and write a method to display its value.
class Product:
    def __init__(self, price):
        self._price = price   # Private attribute

    def display_price(self):
        print("Price: ₹", self._price)

# Create object
product = Product(1500)

# Display price
product.display_price()
```

```python
# 2. Add getter and setter methods for the _price attribute in your Product class to safely access and update the price. Make sure the setter prevents setting a negative price.<br><br><em><strong>Hint:</strong> Raise a ValueError if the new price is less than zero.</em>
class Product:
    def __init__(self, price):
        self._price = price

    # Getter
    def get_price(self):
        return self._price

    # Setter
    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative.")
        self._price = new_price

# Create object
product = Product(1200)

# Get price
print("Current Price:", product.get_price())

# Update price
product.set_price(1500)
print("Updated Price:", product.get_price())

# Negative price (will raise an error)
# product.set_price(-500)
```

```python
# 3. Build a class called Playlist that has a private attribute _songs (a list of song names). Write methods to add a song, remove a song, and get the current list of songs using proper encapsulation.
class Playlist:
    def __init__(self):
        self._songs = []

    # Add song
    def add_song(self, song):
        self._songs.append(song)

    # Remove song
    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
        else:
            print("Song not found.")

    # Get songs
    def get_songs(self):
        return self._songs

# Create object
playlist = Playlist()

# Add songs
playlist.add_song("Kesariya")
playlist.add_song("Perfect")
playlist.add_song("Believer")

print("Playlist:", playlist.get_songs())

# Remove song
playlist.remove_song("Perfect")

print("Updated Playlist:", playlist.get_songs())
```

```python
# 4. Create an abstract class PaymentMethod with an abstract method pay(amount). Then, create two subclasses: UPI and CreditCard, each implementing the pay method with a print statement showing how the payment would be processed.<br><br><em><strong>Hint:</strong> Use the abc module for abstraction.</em>
from abc import ABC, abstractmethod

# Abstract class
class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

# UPI class
class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI.")

# Credit Card class
class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card.")

# Create objects
upi = UPI()
card = CreditCard()

# Make payments
upi.pay(500)
card.pay(1200)
```
