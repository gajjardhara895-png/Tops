```python
# 1. Use open() in write mode to create a file called my_playlist.txt and write the names of 5 songs you listened to this week,
# Open file in write mode
with open("my_playlist.txt", "w") as file:
    file.write("Shape of You\n")
    file.write("Perfect\n")
    file.write("Blinding Lights\n")
    file.write("Believer\n")
    file.write("Levitating\n")

print("Playlist saved successfully!")
```

```python
# 2. Read the my_playlist.txt file you created and print each song name in uppercase using Python file handling.

# Open file in read mode
with open("my_playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```

```python
# Given a JSON file named user_profile.json containing details like username, followers, and bio (similar to an Instagram profile), use the json module to load the file and print the username and number of followers.

import json

# Read JSON file
with open("user_profile.json", "r") as file:
    profile = json.load(file)

print("Username:", profile["username"])
print("Followers:", profile["followers"])
```
