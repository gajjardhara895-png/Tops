```python
# Food items and prices
food_items = ["Pizza", "Burger", "Pasta", "Biryani", "Sandwich"]
prices = [250, 180, 220, 300, 150]

# Print menu
for food, price in zip(food_items, prices):
    print(f"{food} - ₹{price}")
```

```python
# Generator function
def insta_posts_generator(posts):
    for post in posts:
        yield post

# List of Instagram captions
captions = [
    "Enjoying the sunset 🌅",
    "Weekend vibes 😎",
    "Coffee first ☕",
    "Travel diaries ✈️"
]

# Create generator
post_generator = insta_posts_generator(captions)

# Print captions using next()
try:
    while True:
        print(next(post_generator))
except StopIteration:
    print("No more posts.")
```

```python
# Generator function
def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05

# Paytm transaction amounts
transactions = [500, 1200, 750, 3000, 150]

# Print cashback values
for cashback in cashback_generator(transactions):
    print(f"Cashback: ₹{cashback:.2f}")
```
