```python
import json
import os

# ------------------ Order Class ------------------

class Order:
    def __init__(self, order_id, customer_name, items, total_amount, status="Pending"):
        self.order_id = order_id
        self.customer_name = customer_name
        self.items = items
        self.total_amount = total_amount
        self.status = status

    def to_dict(self):
        return {
            "order_id": self.order_id,
            "customer_name": self.customer_name,
            "items": self.items,
            "total_amount": self.total_amount,
            "status": self.status
        }


# ------------------ File Handling ------------------

FILENAME = "orders.json"


def load_orders():
    if not os.path.exists(FILENAME):
        return []

    try:
        with open(FILENAME, "r") as file:
            data = json.load(file)

        return [Order(**order) for order in data]

    except (json.JSONDecodeError, TypeError):
        print("Error reading file. Starting with empty data.")
        return []


def save_orders(orders):
    with open(FILENAME, "w") as file:
        json.dump(
            [order.to_dict() for order in orders],
            file,
            indent=4
        )


# ------------------ Helper Function ------------------

def generate_order_id(orders):
    if not orders:
        return 1

    return max(order.order_id for order in orders) + 1


# ------------------ Place Order ------------------

def place_order(orders):
    print("\n--- Place New Order ---")

    name = input("Enter customer name: ").strip()

    if not name:
        print("Name cannot be empty.")
        return

    items_input = input("Enter items (comma separated): ").strip()

    if not items_input:
        print("Items cannot be empty.")
        return

    items = [item.strip() for item in items_input.split(",")]

    try:
        amount = float(input("Enter total amount: "))

        if amount < 0:
            print("Amount cannot be negative.")
            return

    except ValueError:
        print("Invalid amount. Please enter a numeric value.")
        return

    order_id = generate_order_id(orders)

    new_order = Order(
        order_id,
        name,
        items,
        amount
    )

    orders.append(new_order)

    save_orders(orders)

    print(f"Order {order_id} placed successfully!")


# ------------------ View Orders ------------------

def view_orders(orders):
    print("\n--- All Orders ---")

    if not orders:
        print("No orders found.")
        return

    print(
        f"{'ID':<5}"
        f"{'Name':<15}"
        f"{'Items':<10}"
        f"{'Amount':<12}"
        f"{'Status':<15}"
    )

    print("-" * 57)

    for order in orders:

        status_display = order.status

        if order.status == "Delivered":
            status_display = "Delivered"

        print(
            f"{order.order_id:<5}"
            f"{order.customer_name:<15}"
            f"{len(order.items):<10}"
            f"{order.total_amount:<12.2f}"
            f"{status_display:<15}"
        )


# ------------------ Search Order ------------------

def search_order(orders):
    print("\n--- Search Order ---")

    try:
        oid = int(input("Enter Order ID: "))

    except ValueError:
        print("Invalid Order ID. Please enter a number.")
        return

    for order in orders:

        if order.order_id == oid:

            print("\nOrder Found")
            print("--------------------")
            print(f"ID     : {order.order_id}")
            print(f"Name   : {order.customer_name}")
            print(f"Items  : {', '.join(order.items)}")
            print(f"Amount : Rs {order.total_amount:.2f}")
            print(f"Status : {order.status}")

            if order.status != "Delivered":

                mark = input("Mark as Delivered? (y/n): ").lower()

                if mark == "y":
                    order.status = "Delivered"

                    save_orders(orders)

                    print("Order marked as Delivered.")

            return

    print("Order not found.")


# ------------------ Main Menu ------------------

def main():

    orders = load_orders()

    while True:

        print("\n===== Food Delivery System =====")
        print("1. Place New Order")
        print("2. View All Orders")
        print("3. Search Order by ID")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            place_order(orders)

        elif choice == "2":
            view_orders(orders)

        elif choice == "3":
            search_order(orders)

        elif choice == "4":
            save_orders(orders)
            print("Exiting... Data saved.")
            break

        else:
            print("Invalid choice. Please try again.")


# ------------------ Run Program ------------------

if __name__ == "__main__":
    main()
```
