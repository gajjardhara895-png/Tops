```python
def main():
    items = []
    prices = []

    try:
        n = int(input("Enter number of items: "))
        if n <= 0:
            print("Number of items must be greater than 0.")
            return

        # Input items and prices
        for i in range(n):
            item = input(f"Enter name of item {i+1}: ")

            try:
                price = float(input(f"Enter price of {item}: "))
                if price < 0:
                    print("Price cannot be negative.")
                    return
            except ValueError:
                print("Invalid price input. Please enter a number.")
                return

            items.append(item)
            prices.append(price)

        # Calculate subtotal
        subtotal = sum(prices)

        # GST and delivery
        gst = subtotal * 0.18
        delivery_fee = 30

        # Previous orders input
        try:
            orders = int(input("Enter previous order count: "))
            if orders < 0:
                print("Order count cannot be negative.")
                return
        except ValueError:
            print("Invalid order count. Please enter a number.")
            return

        # Discount
        discount = 0
        if orders > 5:
            discount = subtotal * 0.10

        # Final amount
        final_amount = subtotal + gst + delivery_fee - discount

        # Print Receipt
        print("\n========== RECEIPT ==========")
        for i in range(n):
            print(f"{items[i]:15} Rs {prices[i]:.2f}")

        print("-----------------------------")
        print(f"{'Subtotal':15} Rs {subtotal:.2f}")
        print(f"{'GST (18%)':15} Rs {gst:.2f}")
        print(f"{'Delivery Fee':15} Rs {delivery_fee:.2f}")
        print(f"{'Discount':15} Rs {discount:.2f}")
        print("-----------------------------")
        print(f"{'Final Amount':15} Rs {final_amount:.2f}")
        print("=============================")

    except ValueError:
        print("Invalid input! Please enter correct numeric values.")


# Run program
if __name__ == "__main__":
    main()
```
