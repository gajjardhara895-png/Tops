```python
menu = {
    "Pizza": {"price": 250, "category": "Fast Food"},
    "Burger": {"price": 120, "category": "Fast Food"},
    "Pasta": {"price": 200, "category": "Italian"},
    "Biryani": {"price": 180, "category": "Indian"},
    "Sandwich": {"price": 100, "category": "Snacks"},
    "Coffee": {"price": 80, "category": "Beverage"},
    "Tae": {"price": 50, "category": "Beverage"},
}
while True:
    try:
        options = int(
            input("(0) Exit (1) View all items (2) Search category, (3) Search dish : ")
        )
        if options == 0:
            print("Thank you for wisit")
            break
        elif options == 1:
            print("\n📋 Restaurant Menu:")
            print("*" * 50)
            print(f"{'No.':<5} {'Item':<15} {'Price (₹)':<10} {'Category':<15}")
            print("-" * 50)
            for i, (item, details) in enumerate(menu.items(), start=1):
                print(
                    f"{i:<5} {item:<15} {details['price']:<10} {details['category']:<15}"
                )
        elif options == 2:
            incategory = str(input("Enter a Category name"))
            print("\n📋 Restaurant Menu:")
            print("*" * 50)
            print(f"{'No.':<5} {'Item':<15} {'Price (₹)':<10} {'Category':<15}")
            index = 1
            for no, (item, details) in enumerate(menu.items(), start=1):
                if details["category"] == incategory:
                    print(
                        f"{index:<5} {item:<15} {details['price']:<10} {details['category']:<15}"
                    )
                    index = index + 1
        elif options == 3:
            dish = str(input("input a dish name :"))
            print("\n📋 Restaurant Menu:")
            print("*" * 50)
            print(f"{'No.':<5} {'Item':<15} {'Price (₹)':<10} {'Category':<15}")
            if dish in menu:
                print(
                    f"{"1":<5} {dish:<15} {menu[dish]["price"]:<10} {menu[dish]["category"]}"
                )
            else:
                print("⚠️ Invalid choice! Please try again.")
        else:
            print("⚠️ Invalid choice! Please try again.")
    except ValueError:
        print("⚠️ Invalid choice! Please try again.")
        continue
```
