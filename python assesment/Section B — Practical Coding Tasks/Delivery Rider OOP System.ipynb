```python
import csv
import os

class Rider:
    def __init__(self, rider_id, name, status="Available", total_deliveries=0):
        self.rider_id = rider_id
        self.name = name
        self.status = status
        self.total_deliveries = int(total_deliveries)

    def assign_order(self, order_id):
        if self.status == "On Delivery":
            print(f"❌ Rider {self.name} is already on delivery.")
        else:
            self.status = "On Delivery"
            print(f"✅ Order {order_id} assigned to {self.name}.")

    def complete_delivery(self):
        if self.status == "Available":
            print(f"❌ Rider {self.name} has no active delivery.")
        else:
            self.total_deliveries += 1
            self.status = "Available"
            print(f"✅ Delivery completed by {self.name}.")

    def display_info(self):
        print(f"ID: {self.rider_id}, Name: {self.name}, Status: {self.status}, Total Deliveries: {self.total_deliveries}")



def load_riders(filename):
    riders = []
    if os.path.exists(filename):
        with open(filename, mode="r", newline="") as file:
            reader = csv.reader(file)
            next(reader, None)  # skip header
            for row in reader:
                riders.append(Rider(*row))
    return riders



def save_riders(filename, riders):
    with open(filename, mode="w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["rider_id", "name", "status", "total_deliveries"])
        for rider in riders:
            writer.writerow([rider.rider_id, rider.name, rider.status, rider.total_deliveries])



def main():
    filename = "riders.csv"
    riders = load_riders(filename)

    # If no data, create default riders
    if not riders:
        riders = [
            Rider("1", "Jay"),
            Rider("2", "Rahul"),
            Rider("3", "Amit")
        ]

    while True:
        print("\n===== Delivery Rider System =====")
        print("1. Display Riders")
        print("2. Assign Order")
        print("3. Complete Delivery")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            for rider in riders:
                rider.display_info()

        elif choice == "2":
            rider_id = input("Enter Rider ID: ")
            order_id = input("Enter Order ID: ")

            found = False
            for rider in riders:
                if rider.rider_id == rider_id:
                    rider.assign_order(order_id)
                    found = True
                    break
            if not found:
                print("❌ Rider not found.")

        elif choice == "3":
            rider_id = input("Enter Rider ID: ")

            found = False
            for rider in riders:
                if rider.rider_id == rider_id:
                    rider.complete_delivery()
                    found = True
                    break
            if not found:
                print("❌ Rider not found.")

        elif choice == "4":
            save_riders(filename, riders)
            print("💾 Data saved to riders.csv. Exiting...")
            break

        else:
            print("❌ Invalid choice. Try again.")


# Run program
if __name__ == "__main__":
    main()
```
