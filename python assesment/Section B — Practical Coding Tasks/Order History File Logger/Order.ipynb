```python
import json

while True:
    
    try:
        choice = int(
            input("Enter yuor choice  (1)Enter order (2) view order (3)Exit : ")
        )
        ## exit code 
        if choice == 3:
            break
        ## input order code       
        elif choice == 1:
            name = input("Enter a user name : ")

            items_input = input("Enter items (comma-separated): ")
            items = [item.strip() for item in items_input.split(",")]

            try:
                Amount = int(input("Enter Amount : "))
            except ValueError:
                print("only number allow")
            status = input("Enter order status (Delivered/Pending): ")

            new_order = {
                "Name": name,
                "Items": items,
                "Amount": Amount,
                "Status": status,
            }
            try:
                with open("order_h.json", "r") as file:
                    orders = json.load(file)
            except FileNotFoundError:
                orders = []

            orders.append(new_order)
            with open("order_h.json", "w") as wfile:
                json.dump(orders,wfile,indent=4)
                
              ## view ieder choice  
        elif choice ==2 :
            totalAmount=0
            try:
                with open("order_h.json", "r") as file:
                    orders = json.load(file) 
                    if not orders:
                        print("no order found")
                    else :
                        print(f"{'No':<5} {'Name':<15} {'Items':<20} {'Amount':<10} {'Status':<10}")
                        print("-"*65)
                        for i,o in enumerate(orders,start=1):
                            totalAmount +=o['Amount']
                            print(f"{i:<5} {o['Name']:<15} {', '.join(o['Items']):<20} {o['Amount']:<10} {o['Status']:<10}")
                        print("-"*65)  
                        print(f"Total order is : {len(orders)}     Total amount of all order is : {totalAmount} ") 
            except FileNotFoundError:
                print("File Not Found")           
    except ValueError: 
        print("Invalid opretion")
        continue
```
