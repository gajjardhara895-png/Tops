### **S1 — Data Types & Incorrect Type Impact**

* **Order Total (Rs 499.50)** → `float`
* **Delivery Distance (7.3 km)** → `float`
* **Payment Method ('UPI')** → `str`
* **Restaurant Status (True)** → `bool`

**Explanation:**
Decimal values like 499.50 and 7.3 must be stored as `float` because they contain fractional parts. If we store Rs 499.50 as an `int` (499), we lose the decimal value.

For example, 10% discount:

* Correct: 499.50 × 10% = 49.95
* Wrong (int): 499 × 10% = 49.9

This causes inaccurate billing. So choosing the correct data type ensures precision and correct calculations.

### **S2 — List of Tuples vs Dictionary**

* **List of tuples:**
  `[('Paneer Burger', 180, 'Snacks'), ('Masala Dosa', 90, 'Breakfast')]`

* **Dictionary:**
  `{'Paneer Burger': {'price': 180, 'category': 'Snacks'}}`

**Comparison:**

* Dictionary provides **faster lookup (O(1))** using key (dish name).
* List requires **looping (O(n))** to find a dish.

**Readability:**
Dictionary is more readable:
`menu['Paneer Burger']['price']`
vs
Looping through list to match name.

**Limitation of list:**

* Cannot directly search by name
* Slower for large data

**Dictionary solves this** by giving direct and efficient access.

### **S3 — def vs lambda**

A **lambda function is not suitable** here because:

* It supports only **one expression**
* Cannot handle **multiple conditions clearly**
* Hard to read and debug

A **def function is better** because:

* Supports multiple `if-elif-else` conditions
* More readable and maintainable
* Easier to debug

**Example use of lambda (better case):**
Sorting orders by price:
`sorted(orders, key=lambda x: x['price'])`

Here lambda is perfect because it's simple and short.

### **S4 — JSON Write & Read + Exception Handling**

**Write to JSON:**

```python
import json

order = {"id": 1, "amount": 500, "status": "Delivered"}

with open("orders.json", "w") as file:
    json.dump(order, file)
```

**Read from JSON:**

```python
try:
    with open("orders.json", "r") as file:
        data = json.load(file)
        print(data)
except FileNotFoundError:
    print("File not found, starting fresh.")
    data = {}
```

**Exception:**

* `FileNotFoundError`

**Handling:**
Use `try-except` so the app does not crash and starts with empty data on first run.

### **S5 — Why Class is Better (OOP Design)**

Using a **class** is better because:

* Combines **data + behavior** in one place
* Makes code organized and reusable

**OOP Principles:**

1. **Encapsulation**

   * Data (name, location, status) and methods (update, complete) are inside one class
   * Improves data safety and structure

2. **Abstraction**

   * User interacts with simple methods like `update_location()`
   * Internal logic is hidden

**Benefit:**
Cleaner, scalable, and easier to maintain compared to separate variables or dictionary.

### **S6 — Exception Handling (try-except)**

When user enters text like `"two hundred"`, Python raises:

* **Exception:** `ValueError`

**Solution:**

```python
while True:
    try:
        amount = float(input("Enter order amount: "))
        break
    except ValueError:
        print("Invalid input! Please enter a numeric value.")
```

**Explanation:**

* `try` attempts to convert input
* If invalid, `except` catches error
* Loop ensures user is asked again instead of program crashing

**Result:**
Program becomes user-friendly and robust.
