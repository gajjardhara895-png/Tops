```python
import numpy as np

likes = np.array([100, 150, 200, 120, 180, 250, 300])

comments = np.array([20, 30, 40, 25, 35, 50, 60])

engagement = likes + comments

average_engagement = engagement.mean()

print("Average Engagement:", average_engagement)
```

```python
import numpy as np

prices = np.array([250, 300, 180, 400, 220])

discounts = np.array([50, 70, 30, 100, 40])

final_prices = prices - discounts

print("Final Prices:", final_prices)
```

```python
import numpy as np

scores = np.array([175, 195, 210, 160, 185])

result = scores > 180

print("Scores:", scores)
print("Scores greater than 180:", result)
```

```python
import numpy as np

paytm = np.array([1, 0, 1, 0, 1, 0])
phonepe = np.array([0, 1, 1, 0, 0, 1])

result = np.logical_or(paytm, phonepe)

print("Paid by either app:", result)
```

```python
import numpy as np

steps = np.array([5000, 6500, 7000, 4500, 8000, 6000, 7500])

bonus_steps = steps + 500

total_steps = bonus_steps.sum()

print("Original Steps:", steps)
print("Steps after Bonus:", bonus_steps)
print("Total Steps for the Week:", total_steps)
```
