```python
import sys
import numpy as np

# 1 & 2. Create a Python list and NumPy array

my_list = list(range(1, 1001))
my_array = np.array(range(1, 1001))

# Measure memory usage

list_memory = sys.getsizeof(my_list)
array_memory = my_array.nbytes

print("Memory usage of Python list:", list_memory, "bytes")
print("Memory usage of NumPy array:", array_memory, "bytes")
```

```python
import time
import numpy as np

# 3. Compare addition speed

def compare_addition_speed():
   
    my_list = list(range(10000))
    my_array = np.array(range(10000))

    # Python list: add 5 using a loop/list comprehension
    
    start_time = time.time()

    list_result = [x + 5 for x in my_list]

    list_time = time.time() - start_time

    # NumPy array: vectorized addition
    
    start_time = time.time()

    array_result = my_array + 5

    array_time = time.time() - start_time

   
    print("Python List Time:", list_time, "seconds")
    print("NumPy Array Time:", array_time, "seconds")


compare_addition_speed()
```

```python

import numpy as np

# Restaurant ratings
ratings = np.array([4.2, 3.8, 4.5, 4.0, 3.5])

print("Original Ratings:", ratings)



# 1. Using a For-Loop

# We can multiply every restaurant rating by 2 using a for loop:

result = []

for rating in ratings:
    result.append(rating * 2)

print("Ratings after multiplication:", result)

# 2. Using NumPy Vectorization

# NumPy allows us to multiply the entire array by 2 directly:

result = ratings * 2

print("Ratings after multiplication:", result)
```
