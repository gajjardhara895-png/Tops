```python
import numpy as np

ratings = np.array([4.5, 4.0, np.nan, 3.5, np.nan, 5.0, 4.2])

missing_count = np.isnan(ratings).sum()

print("Missing Ratings:", missing_count)
```

```python
import numpy as np

prices = np.array([50, 150, 500, 1200, 750, 2000, 900])

result = np.clip(prices, 100, 1000)

print(result)
```

```python
import numpy as np

views = np.array([1000, np.nan, 5000, np.inf, 2500, np.nan, np.inf])

finite_values = views[np.isfinite(views)]
max_value = np.max(finite_values)

views[np.isnan(views)] = 0
views[np.isinf(views)] = max_value

print(views)
```

```python
import numpy as np

numbers = np.array([11, 24, 35, 42, 57, 68, 73, 80, 91, 100])

even_indices = np.where(numbers % 2 == 0)

print("Even Number Indices:", even_indices[0])
```
