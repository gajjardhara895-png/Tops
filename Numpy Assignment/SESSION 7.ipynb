```python
import numpy as np

zomato = np.array([4.2, 4.5, 3.8, 4.7, 4.0])
swiggy = np.array([4.0, 4.3, 3.9, 4.6, 4.1])

result = np.concatenate((zomato, swiggy))

print(result)
```

```python
import numpy as np

messages = np.array([
    [20, 15, 30, 25, 18, 22, 28],
    [10, 12, 15, 8, 14, 11, 13],
    [35, 30, 28, 40, 32, 38, 35]
])

new_user = np.array([18, 20, 15, 22, 19, 25, 21])

messages = np.insert(messages, 3, new_user, axis=0)

total_messages = messages.sum(axis=1)

least_user = np.argmin(total_messages)

messages = np.delete(messages, least_user, axis=0)

print(messages)
```

```python
import numpy as np

views = np.array([1000, 2500, 5000, 7500, 10000])

view_array = views.view()
copy_array = views.copy()

view_array[0] = 9999
copy_array[0] = 8888

print("Original:", views)
print("View:", view_array)
print("Copy:", copy_array)
```
