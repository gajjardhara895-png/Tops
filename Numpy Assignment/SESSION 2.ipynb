```python
import numpy as np

# 1. Last 5 Zomato order ratings
my_scores = np.array([4, 5, 3, 4, 2])
print("My Scores:", my_scores)


```

```python
# 2. Even numbers from 10 to 30 (inclusive)
even_numbers = np.arange(10, 31, 2)
print("Even Numbers:", even_numbers)


```

```python
# 3. 8 equally spaced values between 0 and 1
values = np.linspace(0, 1, 8)
print("Linspace Values:", values)
```

```python
# 4. Flipkart-style Add to Cart counter
cart = np.zeros(10, dtype=int)

# Update 3rd and 7th items to 1
# NumPy uses zero-based indexing, so positions 3 and 7 are indices 2 and 6.
cart[2] = 1
cart[6] = 1
print("Cart:", cart)

```

```python
# 5. Paytm-style random OTP codes
np.random.seed(42)
otp_codes = np.random.randint(1000, 10000, 6)
print("OTP Codes:", otp_codes)
```
