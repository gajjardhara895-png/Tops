```python

#1. Movies CSV — loc
import pandas as pd

# Load movie dataset
movies = pd.read_csv("movies.csv")

# Select only title and rating columns using loc
movies.columns = movies.columns.astype(str).str.strip().str.lower()

print(movies.columns.tolist())
```

```python
#2. Flipkart Products — iloc
# # Flipkart products DataFrame
flipkart = pd.DataFrame({
    "product_name": [
        "Laptop", "Mobile", "Headphones", "Keyboard", "Mouse",
        "Monitor", "Tablet", "Smartwatch", "Camera", "Speaker",
        "Printer", "Power Bank"
    ],
    "category": [
        "Electronics", "Electronics", "Audio", "Computer", "Computer",
        "Electronics", "Electronics", "Wearable", "Camera", "Audio",
        "Computer", "Electronics"
    ],
    "price": [
        55000, 25000, 2000, 1500, 800,
        12000, 18000, 5000, 30000, 3500,
        9000, 1200
    ]
})

# Select first 10 rows and all columns
result = flipkart.iloc[:10, :]

print("First 10 Products:")
print(result)
```

```python
# 3. Zomato Orders — Boolean Condition
# Zomato food orders DataFrame
zomato = pd.DataFrame({
    "restaurant": [
        "Dominos", "McDonalds", "Burger King",
        "Subway", "Pizza Hut", "KFC"
    ],
    "order_amount": [
        450, 650, 750, 300, 900, 550
    ],
    "delivery_time": [
        30, 35, 40, 25, 45, 32
    ]
})

# Filter orders where order_amount > 500
result = zomato[zomato["order_amount"] > 500]

print("Orders above 500:")
print(result)
```

```python
#4. Spotify Songs — query()
# # Spotify songs DataFrame
spotify = pd.DataFrame({
    "song": [
        "Song A", "Song B", "Song C",
        "Song D", "Song E"
    ],
    "artist": [
        "Artist 1", "Artist 2", "Artist 3",
        "Artist 4", "Artist 5"
    ],
    "streams": [
        1500000, 800000, 2500000, 1200000, 500000
    ],
    "duration": [
        170, 200, 160, 175, 150
    ]
})

# Filter using query()
result = spotify.query(
    "streams > 1000000 and duration < 180"
)

print("Matching Songs:")
print(result)
```

```python
#5. IPL Matches — Boolean Indexing
# # IPL matches DataFrame
ipl = pd.DataFrame({
    "team1": [
        "MI", "CSK", "RCB", "KKR", "DC"
    ],
    "team2": [
        "CSK", "RCB", "MI", "DC", "KKR"
    ],
    "winner": [
        "MI", "RCB", "CSK", "KKR", "DC"
    ],
    "total_runs": [
        175, 190, 215, 230, 200
    ]
})

# Boolean indexing
result = ipl[
    (ipl["total_runs"] >= 180) &
    (ipl["total_runs"] <= 220)
]

print("Matches with total runs between 180 and 220:")
print(result)
```
