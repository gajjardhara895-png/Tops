```python
#3. Find highest and lowest followers

import pandas as pd

followers = pd.Series([1200, 54000, 780, 150000, 32000])

highest = followers.idxmax()
lowest = followers.idxmin()

print("Highest follower count:", followers[highest])
print("Highest follower index:", highest)

print("Lowest follower count:", followers[lowest])
print("Lowest follower index:", lowest)
```
