```python
#1.Use pd.concat to combine two DataFrames representing January and February orders from a Swiggy-style food delivery app, stacking them row-wise to create a single DataFrame with all orders.
import pandas as pd

# January orders
january = pd.DataFrame({
    "order_id": [101, 102, 103],
    "customer": ["Dhara", "Riya", "Neha"],
    "amount": [250, 400, 300]
})

# February orders
february = pd.DataFrame({
    "order_id": [104, 105, 106],
    "customer": ["Amit", "Priya", "Rahul"],
    "amount": [500, 350, 450]
})

# Combine row-wise
all_orders = pd.concat([january, february], ignore_index=True)

print("All Orders:")
print(all_orders)
```

```python
#2.Given two DataFrames — one with Flipkart product details and another with their respective ratings — use pd.concat to merge them column-wise so each product row includes its rating.
import pandas as pd

# Flipkart product details
products = pd.DataFrame({
    "product_id": [101, 102, 103, 104, 105],
    "product_name": ["Laptop", "Mobile", "Headphones", "Smartwatch", "Keyboard"],
    "price": [50000, 20000, 2000, 3000, 1500]
})

# Product ratings
ratings = pd.DataFrame({
    "rating": [4.5, 4.2, 4.0, 4.3, 4.1]
})

# Combine column-wise
result = pd.concat([products, ratings], axis=1)

print("Products with Ratings:")
print(result)
```

```python
#3.Create a DataFrame showing daily song streams for three users on Spotify for one week, then use the pivot function to reshape the data so each row is a user and each column is a day.
import pandas as pd

# Daily song streams for 3 users for one week
spotify = pd.DataFrame({
    "user": [
        "Dhara", "Dhara", "Dhara", "Dhara", "Dhara", "Dhara", "Dhara",
        "Riya", "Riya", "Riya", "Riya", "Riya", "Riya", "Riya",
        "Neha", "Neha", "Neha", "Neha", "Neha", "Neha", "Neha"
    ],
    "day": [
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"
    ],
    "streams": [
        120, 150, 100, 180, 200, 250, 220,
        100, 130, 160, 140, 180, 210, 190,
        80, 120, 110, 150, 170, 200, 180
    ]
})

# Pivot the data
result = spotify.pivot(
    index="user",
    columns="day",
    values="streams"
)

print("Pivoted Spotify Data:")
print(result)
```

```python
#4.Given a DataFrame of Zomato restaurant orders (with columns: restaurant, month, total_amount), use pivot_table to calculate the average order amount per restaurant for each month.
import pandas as pd

# Zomato restaurant orders
orders = pd.DataFrame({
    "restaurant": [
        "Dominos", "Dominos", "Dominos",
        "Pizza Hut", "Pizza Hut", "Pizza Hut",
        "Subway", "Subway", "Subway"
    ],
    "month": [
        "January", "February", "March",
        "January", "February", "March",
        "January", "February", "March"
    ],
    "total_amount": [
        500, 600, 700,
        400, 550, 650,
        300, 450, 500
    ]
})

# Calculate average order amount
result = pd.pivot_table(
    orders,
    index="restaurant",
    columns="month",
    values="total_amount",
    aggfunc="mean"
)

print("Average Order Amount:")
print(result)
```

```python
#5. Download three monthly sales CSV files for a mock Myntra store (e.g., sales_jan.csv, sales_feb.csv, sales_mar.csv), then use pd.concat to combine them into one DataFrame and display the total sales per product.<br><br><em><strong>Hint:</strong> Use groupby on the product column after concatenation to sum sales.</em>
import pandas as pd

# Create January sales
jan = pd.DataFrame({
    "product": ["Kurti", "Jeans", "Tshirt"],
    "sales": [5000, 4000, 3000]
})

# Create February sales
feb = pd.DataFrame({
    "product": ["Kurti", "Jeans", "Tshirt"],
    "sales": [6000, 5000, 3500]
})

# Create March sales
mar = pd.DataFrame({
    "product": ["Kurti", "Jeans", "Tshirt"],
    "sales": [7000, 4500, 4000]
})

# Save as CSV files
jan.to_csv("sales_jan.csv", index=False)
feb.to_csv("sales_feb.csv", index=False)
mar.to_csv("sales_mar.csv", index=False)

# Read CSV files
jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")

# Combine all files
all_sales = pd.concat([jan, feb, mar], ignore_index=True)

print("Combined Sales:")
print(all_sales)

# Total sales per product
total_sales = all_sales.groupby("product")["sales"].sum()

print("\nTotal Sales Per Product:")
print(total_sales)
```
