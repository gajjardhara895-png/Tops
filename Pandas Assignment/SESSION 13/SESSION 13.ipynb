```python
#2.Perform univariate analysis on the 'runs' column: calculate and print the mean, median, mode, minimum, and maximum runs scored by batsmen in the IPL_2023_Batsmen dataset.
import pandas as pd

# Load dataset
ipl = pd.read_csv("IPL_2023_Batsmen.csv")

# Select runs column
runs = ipl["runs"]

# Calculate statistics
mean_runs = runs.mean()
median_runs = runs.median()
mode_runs = runs.mode()
min_runs = runs.min()
max_runs = runs.max()

# Print results
print("Mean Runs:", mean_runs)
print("Median Runs:", median_runs)
print("Mode Runs:", mode_runs.tolist())
print("Minimum Runs:", min_runs)
print("Maximum Runs:", max_runs)
```

```python
#3.Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.
import pandas as pd

# Create DataFrame
zomato = pd.DataFrame({
    "name": [
        "Dominos",
        "Pizza Hut",
        "Subway",
        "Honest",
        "La Pino's Pizza"
    ],
    "rating": [4.2, 4.0, 4.1, 4.5, 4.3],
    "votes": [1200, 900, 1100, 1500, 1300],
    "avg_cost": [500, 600, 400, 350, 550]
})

print("Zomato Restaurants:")
print(zomato)

# Calculate correlation
correlation = zomato["rating"].corr(zomato["votes"])

print("\nCorrelation between Rating and Votes:", correlation)
```
