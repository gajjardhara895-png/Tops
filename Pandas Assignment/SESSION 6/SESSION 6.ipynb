```python
#1.Load a CSV file of Spotify song streams (with columns: song, artist, streams, date) into a pandas DataFrame and use drop_duplicates() to remove any duplicate song entries, keeping only the first occurrence.
import pandas as pd

# Load CSV file
spotify = pd.read_csv("spotify_streams.csv")

print("Before removing duplicates:")
print(spotify)

# Remove duplicate song entries
spotify_cleaned = spotify.drop_duplicates(
    subset=["song"],
    keep="first"
)

print("\nAfter removing duplicates:")
print(spotify_cleaned)
```

```python
#2.Given a pandas DataFrame of Flipkart product reviews with columns: product_id, user_id, rating, review_text, identify and print the number of duplicate reviews (where product_id and user_id are both the same) before and after using drop_duplicates().

import pandas as pd

reviews = pd.DataFrame({
    "product_id": [101, 102, 101, 103, 102, 101],
    "user_id": [1, 2, 1, 3, 2, 4],
    "rating": [5, 4, 5, 3, 4, 2],
    "review_text": [
        "Excellent",
        "Good product",
        "Excellent",
        "Average",
        "Good product",
        "Nice product"
    ]
})

# Number of duplicate reviews before removing
duplicate_count = reviews.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Duplicate reviews before:", duplicate_count)

# Remove duplicates
cleaned_reviews = reviews.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)

# Number of duplicates after removing
duplicate_after = cleaned_reviews.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Duplicate reviews after:", duplicate_after)

print("\nCleaned DataFrame:")
print(cleaned_reviews)
```

```python
#3.Use the IQR method to detect outliers in the 'order_amount' column of a Zomato orders DataFrame and print the order IDs that are considered outliers.<br><br><em><strong>Hint:</strong> Calculate Q1, Q3, and IQR, then filter orders outside the [Q1 - 1.5*IQR, Q3 + 1.5*IQR] range.</em>

import pandas as pd

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "order_amount": [250, 300, 275, 320, 290, 350, 2000]
})

# Calculate Q1 and Q3
Q1 = orders["order_amount"].quantile(0.25)
Q3 = orders["order_amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Find outliers
outliers = orders[
    (orders["order_amount"] < lower_limit) |
    (orders["order_amount"] > upper_limit)
]

print("\nOutlier Order IDs:")
print(outliers["order_id"])
```

```python
#4.For a DataFrame containing daily IPL match ticket sales (columns: date, tickets_sold), calculate the Z-score for each day's tickets_sold, and print the dates where the absolute Z-score is greater than 2.<br><br><em><strong>Constraint:</strong> Use pandas and scipy.stats.zscore for calculations.</em>

import pandas as pd
from scipy.stats import zscore

ipl = pd.DataFrame({
    "date": [
        "2026-04-01",
        "2026-04-02",
        "2026-04-03",
        "2026-04-04",
        "2026-04-05",
        "2026-04-06"
    ],
    "tickets_sold": [
        5000,
        5200,
        5100,
        5300,
        5000,
        15000
    ]
})

# Calculate Z-score
ipl["z_score"] = zscore(ipl["tickets_sold"])

print(ipl)

# Find rows where absolute Z-score > 2
outliers = ipl[ipl["z_score"].abs() > 2]

print("\nOutlier Dates:")
print(outliers[["date", "tickets_sold", "z_score"]])
```

```python
#5.Use ChatGPT or Copilot to generate a pandas function that finds and returns all outlier transaction amounts in a Paytm wallet DataFrame using the Z-score method, then test the function with sample data.

import pandas as pd
from scipy.stats import zscore

def find_outliers(df, column, threshold=2):
    df = df.copy()
    df["z_score"] = zscore(df[column])
    
    return df[df["z_score"].abs() > threshold]


paytm = pd.DataFrame({
    "transaction_id": [101, 102, 103, 104, 105, 106],
    "transaction_amount": [100, 150, 120, 130, 110, 5000]
})

result = find_outliers(
    paytm,
    "transaction_amount"
)

print(result)
```
