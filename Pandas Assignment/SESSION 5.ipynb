```python
#1. Download a CSV file of IPL cricket players with columns for player name, team, and age, then use pandas to load it and print out all rows where the age column has missing values.

import pandas as pd

# Create sample IPL player data
ipl = pd.DataFrame({
    "player_name": [
        "Virat Kohli",
        "Rohit Sharma",
        "Jasprit Bumrah",
        "Shubman Gill",
        "Ravindra Jadeja",
        "Hardik Pandya"
    ],
    "team": [
        "RCB",
        "MI",
        "MI",
        "GT",
        "CSK",
        "MI"
    ],
    "age": [
        37,
        None,
        32,
        None,
        37,
        32
    ]
})

# Display rows where age is missing
result = ipl[ipl["age"].isna()]

print(result)
```

```python
#2. Using a dataset of food delivery orders (with columns: order_id, customer_name, delivery_rating), use pandas to drop all rows where the delivery_rating is missing and display the cleaned DataFrame.

import pandas as pd

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer_name": ["Dhara", "Riya", "Neha", "Priya", "Kajal"],
    "delivery_rating": [5, None, 4, None, 3]
})

# Drop rows where delivery_rating is missing
cleaned_orders = orders.dropna(subset=["delivery_rating"])

print(cleaned_orders)
```

```python
#3. Given a Flipkart-style product reviews dataset with some missing 'rating' values, use pandas to fill all missing ratings with the mean rating of the dataset and print the updated DataFrame. Use the fillna() method with the mean() function.

import pandas as pd

products = pd.DataFrame({
    "product_name": [
        "Mobile",
        "Laptop",
        "Headphones",
        "Smartwatch",
        "Keyboard"
    ],
    "rating": [4.5, 4.0, None, 5.0, None]
})

# Calculate mean rating
mean_rating = products["rating"].mean()

# Fill missing ratings with mean
products["rating"] = products["rating"].fillna(mean_rating)

print(products)
```

```python
#4. You have a Spotify playlist CSV with missing 'duration' values for some songs. Replace all missing durations with the median duration of the available songs using pandas, then save the cleaned DataFrame to a new CSV file.

import pandas as pd

spotify = pd.DataFrame({
    "song": [
        "Kesariya",
        "Tum Hi Ho",
        "Apna Bana Le",
        "Chaleya",
        "Heeriye"
    ],
    "duration": [
        268,
        None,
        245,
        None,
        220
    ]
})

# Calculate median duration
median_duration = spotify["duration"].median()

# Fill missing durations with median
spotify["duration"] = spotify["duration"].fillna(median_duration)

print(spotify)

# Save cleaned DataFrame
spotify.to_csv("spotify_cleaned.csv", index=False)
```
